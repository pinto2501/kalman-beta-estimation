```python
def get_stock_price_data(
    output_dir, 
    stock_list, 
    from_date=START,
    to_date=END,
):

    df = yf.download(stock_list, start=from_date, end=to_date, interval="1d")
    df_stacked = df.stack(level="Ticker", future_stack=True).reset_index()
    os.makedirs(output_dir, exist_ok=True)

    cols = ["Date", "Close", "High", "Low", "Open", "Volume"]

    for stock in stock_list:
        file_path = os.path.join(output_dir, f"{stock}.csv")
        data = df_stacked[df_stacked['Ticker'] == stock].loc[:, cols]
        data.to_csv(file_path, index=False)
```

```python
get_stock_price_data(output_dir="../data", stock_list=MEGACAPS+[BENCHMARK])
```
